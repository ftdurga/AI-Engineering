#### Chat Completions API

1. The simplest way to call an LLM
2. It's called Chat Completions because it's saying: "here is a conversation, please predict what should come next"
3. The Chat Completions API was invented by OpenAI, but it's so popular that everybody uses it!

```python
import os
from dotenv import load_dotenv

load_dotenv(override=True)
api_key = os.getenv('API_KEY')

if not api_key:
    print("No API key was found")
else:
    print("API key found and looks good so far!")
```

#### OpenAI Package

It's known as a Python Client Library. It's nothing more than a wrapper around making this exact call to the HTTP endpoint.

It just allows you to work with nice Python code instead of messing around with janky JSON objects. But that's it. It's open-source and lightweight. Some people think it contains OpenAI model code - it doesn't!

#### Ollama also gives an OpenAI compatible endpoint!

```python
from openai import OpenAI

ollama = OpenAI(base_url=os.getenv("BASE_URL"), api_key=os.getenv("API_KEY"))
```

```python
# Get a fun fact
response = ollama.chat.completions.create(model=os.getenv("MODEL"), messages=[{"role": "user", "content": "Tell me a fun fact"}])
response.choices[0].message.content
```
